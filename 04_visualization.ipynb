```python
import pandas as pd
import matplotlib.pyplot as plt
import os

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../data/raw/ev_charging_sessions_raw.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
df["start_time"] = pd.to_datetime(df["start_time"])

df["date"] = df["start_time"].dt.date

print("Date conversion completed!")

print(df[["start_time", "date"]].head())
```

```python
os.makedirs("../visuals", exist_ok=True)

print("Visuals folder is ready!")
```

```python
city_sessions = df["city"].value_counts()

plt.figure(figsize=(10, 6))

city_sessions.plot(kind="bar")

plt.title("EV Charging Sessions by City")
plt.xlabel("City")
plt.ylabel("Number of Charging Sessions")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../visuals/chart1_city_comparison.png", dpi=300)

plt.show()
```

Figure 1: EV Charging Sessions by City

The bar chart compares the number of EV charging sessions across different Kenyan cities. It helps identify which cities have the highest and lowest charging activity. Cities with more sessions may have greater demand for charging services.
