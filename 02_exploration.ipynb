```python
import pandas as pd

df = pd.read_csv("../data/raw/ev_charging_sessions_raw.csv")

print("Shape:", df.shape)
df.head(10)
```

```python
print(df.tail(5))
print()
df.info()
```

```python
print(df.dtypes)
df.describe(include="all")
```

```python
missing = df.isna().sum()
percent = (missing / len(df) * 100).round(2)

summary = pd.DataFrame({"missing": missing, "percent": percent})
summary[summary["missing"] > 0].sort_values("missing", ascending=False)
```

```python
print("Duplicate rows:", df.duplicated().sum())
df[df.duplicated(keep=False)].sort_values("session_id").head(6)
```

```python
print("Negative energy:", (df[df["energy_kwh"] < 0]).shape[0])
print("Zero cost:", (df["cost_kes"] == 0).sum())
print("Maximum energy:", df["energy_kwh"].max())
print()
print("Highest energy values:")
print(df.sort_values("energy_kwh", ascending=False)[["session_id", "vehicle_type", "energy_kwh"]].head(8))
```
