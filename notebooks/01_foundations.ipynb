```python
import pandas as pd
print(pd.__version__)
```

```python
df = pd.read_csv("../data/raw/ev_charging_sessions_raw.csv")
print(df.shape)
```

```python
print(df.shape)
print(df.columns)
df.head()
```

```python
# 1. A variable
total_stations = 48
print("Number of stations:", total_stations)

# 2. A list
cities = ["Nairobi", "Mombasa", "Kisumu", "Nakuru", "Eldoret"]
print("Cities:", cities)
```

```python
# 3. A dictionary (labels paired with values)
stations_per_city = {
    "Nairobi": 28,
    "Mombasa": 7,
    "Kisumu": 4,
    "Nakuru": 5,
    "Eldoret": 4
}
print("Stations in Nairobi:", stations_per_city["Nairobi"])

# 4. A loop (repeat an action for each item)
for city in cities:
    print(city, "has", stations_per_city[city], "stations")
```

```python
# 5. A conditional (make a decision)
rows = df.shape[0]

if rows >= 15000:
    print("Meets the minimum of 15,000 rows. Rows found:", rows)
else:
    print("Not enough rows. Rows found:", rows)
```

```python
# 6. Function one: average energy for one vehicle type
def average_energy(vehicle):
    subset = df[df["vehicle_type"] == vehicle]
    return round(subset["energy_kwh"].mean(), 2)

print("Average kWh for Electric Car:", average_energy("Electric Car"))
print("Average kWh for Electric Motorbike:", average_energy("Electric Motorbike"))


# 7. Function two: check if a dataset meets the project size rule
def meets_requirement(data, min_rows=15000, min_cols=10):
    rows, cols = data.shape
    if rows >= min_rows and cols >= min_cols:
        return True
    return False

print("Meets 15,000 x 10 requirement:", meets_requirement(df))
```
